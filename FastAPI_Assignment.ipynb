```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "FastAPI Assignment Running"}
```

```python
@app.get("/")
def welcome():
    return {"message": "Welcome to FastAPI Assignment"}
```

```python
@app.get("/student")
def student():
    return {
        "name": "Varnitha",
        "branch": "CSE",
        "year": "3rd Year"
    }
```

```python
@app.get("/add/{a}/{b}")
def add_numbers(a: int, b: int):
    result = a + b
    return {"number1": a, "number2": b, "sum": result}
```

```python
@app.get("/multiply/{a}/{b}")
def multiply_numbers(a: int, b: int):
    result = a * b
    return {"number1": a, "number2": b, "product": result}
```

```python
@app.get("/hello/{name}")
def say_hello(name: str):
    return {"message": f"Hello {name}, welcome to FastAPI"}
```
