```python
# Smart Light Controller Problem:

class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"  # Default state is off

    def set_status(self, action):
        # We update the light's status based on the user's command
        self.status = action
        print(f"{self.name} is {self.status}")

# Human Logic: Create the light and flip the switch
my_light = SmartLight("Bedroom Light")
my_light.set_status("ON")
```

```python
# Employee ID Card System Problem:

class Employee:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_card(self):
        # Formatting the output to look like a physical card
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")

# Human Logic: Register Rahul and print his card
emp1 = Employee("Rahul", "EMP102", "IT")
emp1.display_card()
```

```python
# Mobile Contact Record Problem:

class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def show_contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")

# Human Logic: Save Anita's number to the address book
new_contact = Contact("Anita", "9876543210")
new_contact.show_contact()
```

```python
# Product Price Tag Generator Problem:

class PriceTag:
    def __init__(self, product_name, price):
        self.product_name = product_name
        self.price = price

    def print_tag(self):
        # Adding the currency symbol '₹' to the numeric price
        print(f"Product: {self.product_name}")
        print(f"Price: ₹{self.price}")

# Human Logic: Generate a tag for a shelf item
item = PriceTag("Headphones", 2499)
item.print_tag()
```

```python
# Movie Rating Display System Problem:

class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display_details(self):
        # Showing the rating relative to a max score of 5
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating} / 5")

# Human Logic: Check the score of a movie on a streaming app
film = Movie("Inception", 4.8)
film.display_details()
```

```python
# Delivery Address Manager Problem:

class DeliveryManager:
    def __init__(self, customer, address):
        self.customer = customer
        self.address = address

    def print_label(self):
        print("Delivery Details")
        print(f"Customer: {self.customer}")
        print(f"Address: {self.address}")

# Human Logic: Prepare a package for courier dispatch
delivery = DeliveryManager("Suman", "Hyderabad")
delivery.print_label()
```
