```python
# Smart Parking Lot Management System problem:

def manage_parking(capacity, logs):
    current_count = 0
    peak_usage = 0

    for action in logs:
        if action == "IN":
            current_count += 1
        elif action == "OUT":
            current_count -= 1
        
        # Track the highest number of cars seen at once
        if current_count > peak_usage:
            peak_usage = current_count

    # Determine status and alerts
    status = "Full" if current_count >= capacity else "Available"
    if current_count > capacity:
        print("ALERT: Capacity Exceeded!")

    print(f"Currently Parked Vehicles: {current_count}")
    print(f"Peak Usage Today: {peak_usage}")
    print(f"Parking Status: {status}")

# Sample Input
manage_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
# Online Food Delivery Time Estimator problem:

def estimate_delivery(distance, traffic, weather):
    # Base time: 5 minutes per kilometer
    eta = distance * 5
    
    # Traffic Adjustments
    if traffic == "High":
        eta += 20
    elif traffic == "Medium":
        eta += 10
        
    # Weather Adjustments
    if weather == "Rainy":
        eta += 15
    elif weather == "Snowy":
        eta += 25

    print(f"Estimated Delivery Time: {eta} minutes")

# Sample Input
estimate_delivery(8, "High", "Rainy")
```

```python
# Movie Theatre Seat Occupancy Analyzer problem:

def analyze_occupancy(total_seats, booked_seats_list):
    booked_count = len(booked_seats_list)
    occupancy_pct = (booked_count / total_seats) * 100
    
    status = "Available"
    if occupancy_pct == 100:
        status = "Housefull"
    elif occupancy_pct >= 70:
        status = "Almost Full"
        
    print(f"Occupancy: {occupancy_pct}%")
    print(f"Show Status: {status}")
    
    if occupancy_pct > 80:
        print("Suggestion: Open additional shows.")

# Sample Input (150 entries in list)
analyze_occupancy(200, [1] * 150)
```

```python
# Cloud Server Load Classification System problem:

def classify_server_load(readings):
    if not readings: return
    
    avg_load = sum(readings) / len(readings)
    
    # Classification Logic
    if avg_load < 50:
        status = "Normal"
    elif 50 <= avg_load <= 80:
        status = "Warning"
    else:
        status = "Critical"
        
    print(f"Average CPU Load: {avg_load}%")
    print(f"Server Status: {status}")

# Sample Input
classify_server_load([45, 60, 70, 85, 90])
```

```python
# Smart Classroom Resource Usage Monitor problem:

def monitor_resources(usage_dict):
    overused = []
    threshold = 8 # Define overuse as > 8 hours
    
    for resource, hours in usage_dict.items():
        if hours >= threshold:
            overused.append(resource)
            
    print(f"Overused Resources: {', '.join(overused) if overused else 'None'}")
    print(f"Energy Alert: {'Yes' if overused else 'No'}")

# Sample Input
resource_data = {"Projector": 6, "AC": 9, "Lights": 4}
monitor_resources(resource_data)
```

```python
# Online Event Registration Capacity Controller problem:

def manage_registrations(capacity, total_registrations):
    confirmed = min(capacity, total_registrations)
    waitlisted = max(0, total_registrations - capacity)
    
    status = "Closed" if total_registrations >= capacity else "Open"
    
    print(f"Confirmed Registrations: {confirmed}")
    print(f"Waitlisted Users: {waitlisted}")
    print(f"Registration Status: {status}")

# Sample Input
manage_registrations(100, 105)
```
